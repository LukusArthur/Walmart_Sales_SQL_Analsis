```python
#Check it again 
df.shape
```

```python
#For the Unit Price Column, we can't calculate unless it is integer / float 
#The thing is "it has $ sign in it , apart from that we can't do nth else than converting that to space or some chars"ArithmeticError
df['unit_price'] = df['unit_price'].str.replace('$','').astype(float) #1st type -Convert the $ Sign into sth else with str.replace
#then change the data type 
```

```python
#We Add A New Column Called "Total Amount" cause there is no in our dataframe 

df.columns
```

```python
df['Total Amount'] = df['unit_price'] * df['quantity']
```

```python
#We all done with Filtering so we will convert it into Csc file from our dataframe 

df.to_csv('Walmart_Clean_Data.csv', index = False)
```

```python
#Now Make The  Connection with PostgreSQL 
engine_psql = create_engine("postgresql+psycopg2://postgres:123456@localhost:5432/Walmart_db")

try: 
    engine_psql
    print("Connection Successed to PSQL")
except:
    print("Unable To Connect")
```

```python
#then put inside our PostgreSQL as a table 

df.to_sql(name = 'Walmart', con = engine_psql, if_exists= 'append', index = False)
```

```python
df.to_sql('walmart_sales', con=engine_psql, if_exists='replace', index=False)
```
